```python
import numpy as np
import pandas as pd

from sklearn.metrics import (
    roc_auc_score,
    confusion_matrix
)

from sklearn.utils import resample

# =====================================================
# Cutoffs (prevalência da classe positiva)
# =====================================================

cutoff1 = y_train1.mean()
cutoff2 = y_train2.mean()
cutoff4 = y_train4.mean()

print(f"Cutoff Etapa 1: {cutoff1:.4f}")
print(f"Cutoff Etapa 2: {cutoff2:.4f}")
print(f"Cutoff Etapa 4: {cutoff4:.4f}")

# =====================================================
# Função para calcular métricas + IC95%
# =====================================================

def calcular_metricas(modelo, X_test, y_test, cutoff, n_boot=2000):

    # Probabilidades
    y_prob = modelo.predict_proba(X_test)[:, 1]

    # ------------------------------------
    # Métricas observadas
    # ------------------------------------
    auc = roc_auc_score(y_test, y_prob)

    y_pred = (y_prob >= cutoff).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_test,
        y_pred,
        labels=[0, 1]
    ).ravel()

    sens = tp / (tp + fn)
    esp = tn / (tn + fp)
    vpp = tp / (tp + fp) if (tp + fp) > 0 else np.nan
    vpn = tn / (tn + fn) if (tn + fn) > 0 else np.nan

    # ------------------------------------
    # Bootstrap
    # ------------------------------------

    aucs = []
    sens_boot = []
    esp_boot = []
    vpp_boot = []
    vpn_boot = []

    y_test_np = np.asarray(y_test)
    y_prob_np = np.asarray(y_prob)

    for _ in range(n_boot):

        idx = resample(
            np.arange(len(y_test_np)),
            replace=True,
            n_samples=len(y_test_np)
        )

        y_true_b = y_test_np[idx]
        y_prob_b = y_prob_np[idx]

        if len(np.unique(y_true_b)) < 2:
            continue

        aucs.append(
            roc_auc_score(y_true_b, y_prob_b)
        )

        y_pred_b = (y_prob_b >= cutoff).astype(int)

        tn_b, fp_b, fn_b, tp_b = confusion_matrix(
            y_true_b,
            y_pred_b,
            labels=[0, 1]
        ).ravel()

        sens_boot.append(
            tp_b / (tp_b + fn_b)
            if (tp_b + fn_b) > 0 else np.nan
        )

        esp_boot.append(
            tn_b / (tn_b + fp_b)
            if (tn_b + fp_b) > 0 else np.nan
        )

        vpp_boot.append(
            tp_b / (tp_b + fp_b)
            if (tp_b + fp_b) > 0 else np.nan
        )

        vpn_boot.append(
            tn_b / (tn_b + fn_b)
            if (tn_b + fn_b) > 0 else np.nan
        )

    # ------------------------------------
    # IC95%
    # ------------------------------------

    def ic95(x):
        return np.nanpercentile(x, [2.5, 97.5])

    return {

        "AUC": auc,
        "AUC_LI": ic95(aucs)[0],
        "AUC_LS": ic95(aucs)[1],

        "Sens": sens,
        "Sens_LI": ic95(sens_boot)[0],
        "Sens_LS": ic95(sens_boot)[1],

        "Esp": esp,
        "Esp_LI": ic95(esp_boot)[0],
        "Esp_LS": ic95(esp_boot)[1],

        "VPP": vpp,
        "VPP_LI": ic95(vpp_boot)[0],
        "VPP_LS": ic95(vpp_boot)[1],

        "VPN": vpn,
        "VPN_LI": ic95(vpn_boot)[0],
        "VPN_LS": ic95(vpn_boot)[1],

        # Matriz de confusão
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp
    }

# =====================================================
# Aplicar aos modelos
# =====================================================

res1 = calcular_metricas(
    modelo=xgb_novo1,
    X_test=X_test_novo1,
    y_test=y_test1,
    cutoff=cutoff1
)

res2 = calcular_metricas(
    modelo=cat_novo2,
    X_test=X_test_novo2,
    y_test=y_test2,
    cutoff=cutoff2
)

res4 = calcular_metricas(
    modelo=cat_novo4,
    X_test=X_test_novo4,
    y_test=y_test4,
    cutoff=cutoff4
)

# =====================================================
# Resultados
# =====================================================

resultados = [
    ("Etapa 1", cutoff1, res1),
    ("Etapa 2", cutoff2, res2),
    ("Etapa 4", cutoff4, res4)
]

for nome, cutoff, res in resultados:

    print("\n" + "=" * 70)
    print(nome)
    print("=" * 70)

    print(f"Cutoff utilizado : {cutoff:.4f}")

    print(
        f"AUC-ROC        : {res['AUC']:.4f} "
        f"({res['AUC_LI']:.4f} - {res['AUC_LS']:.4f})"
    )

    print(
        f"Sensibilidade  : {res['Sens']:.4f} "
        f"({res['Sens_LI']:.4f} - {res['Sens_LS']:.4f})"
    )

    print(
        f"Especificidade : {res['Esp']:.4f} "
        f"({res['Esp_LI']:.4f} - {res['Esp_LS']:.4f})"
    )

    print(
        f"VPP            : {res['VPP']:.4f} "
        f"({res['VPP_LI']:.4f} - {res['VPP_LS']:.4f})"
    )

    print(
        f"VPN            : {res['VPN']:.4f} "
        f"({res['VPN_LI']:.4f} - {res['VPN_LS']:.4f})"
    )

    # ===========================================
    # Matriz de confusão
    # ===========================================

    print("\nMatriz de Confusão")

    cm = pd.DataFrame(
        [
            [res["TN"], res["FP"]],
            [res["FN"], res["TP"]]
        ],
        index=["Real 0", "Real 1"],
        columns=["Predito 0", "Predito 1"]
    )

    print(cm)

    print("\nValores individuais:")
    print(f"TN = {res['TN']}")
    print(f"FP = {res['FP']}")
    print(f"FN = {res['FN']}")
    print(f"TP = {res['TP']}")
```

```python
import numpy as np
import pandas as pd

from sklearn.metrics import (
    roc_auc_score,
    confusion_matrix,
    brier_score_loss
)

from sklearn.utils import resample

import statsmodels.api as sm


# =====================================================
# Função para calcular métricas + IC95%
# =====================================================

def calcular_metricas(modelo, X_test, y_test, cutoff, n_boot=2000):

    # ------------------------------------
    # Probabilidades
    # ------------------------------------
    y_prob = modelo.predict_proba(X_test)[:, 1]

    y_test_np = np.asarray(y_test)
    y_prob_np = np.asarray(y_prob)

    # ------------------------------------
    # Métricas observadas
    # ------------------------------------

    auc = roc_auc_score(
        y_test_np,
        y_prob_np
    )

    brier = brier_score_loss(
        y_test_np,
        y_prob_np
    )

    y_pred = (y_prob_np >= cutoff).astype(int)


    tn, fp, fn, tp = confusion_matrix(
        y_test_np,
        y_pred,
        labels=[0, 1]
    ).ravel()


    sens = tp / (tp + fn)
    esp = tn / (tn + fp)

    vpp = (
        tp / (tp + fp)
        if (tp + fp) > 0 else np.nan
    )

    vpn = (
        tn / (tn + fn)
        if (tn + fn) > 0 else np.nan
    )


    # ------------------------------------
    # Calibração observada
    # ------------------------------------

    eps = 1e-15

    p = np.clip(
        y_prob_np,
        eps,
        1-eps
    )

    logit_p = np.log(
        p/(1-p)
    )


    # intercept

    modelo_intercept = sm.GLM(
        y_test_np,
        np.ones(len(logit_p)),
        family=sm.families.Binomial(),
        offset=logit_p
    ).fit()


    calibration_intercept = (
        modelo_intercept.params[0]
    )


    # slope

    X_cal = sm.add_constant(logit_p)


    modelo_cal = sm.GLM(
        y_test_np,
        X_cal,
        family=sm.families.Binomial()
    ).fit()


    calibration_slope = (
        modelo_cal.params[1]
    )


    # ------------------------------------
    # Bootstrap
    # ------------------------------------

    aucs = []

    sens_boot = []
    esp_boot = []

    vpp_boot = []
    vpn_boot = []

    brier_boot = []

    intercept_boot = []
    slope_boot = []


    for _ in range(n_boot):

        idx = resample(
            np.arange(len(y_test_np)),
            replace=True,
            n_samples=len(y_test_np)
        )


        y_true_b = y_test_np[idx]
        y_prob_b = y_prob_np[idx]


        # precisa das duas classes
        if len(np.unique(y_true_b)) < 2:
            continue


        # ----------------------------
        # AUC
        # ----------------------------

        aucs.append(
            roc_auc_score(
                y_true_b,
                y_prob_b
            )
        )


        # ----------------------------
        # Brier
        # ----------------------------

        brier_boot.append(
            brier_score_loss(
                y_true_b,
                y_prob_b
            )
        )


        # ----------------------------
        # Calibração
        # ----------------------------

        p_b = np.clip(
            y_prob_b,
            eps,
            1-eps
        )

        logit_b = np.log(
            p_b/(1-p_b)
        )


        try:

            modelo_i = sm.GLM(
                y_true_b,
                np.ones(len(logit_b)),
                family=sm.families.Binomial(),
                offset=logit_b
            ).fit()


            intercept_boot.append(
                modelo_i.params[0]
            )


            X_b = sm.add_constant(
                logit_b
            )


            modelo_s = sm.GLM(
                y_true_b,
                X_b,
                family=sm.families.Binomial()
            ).fit()


            slope_boot.append(
                modelo_s.params[1]
            )


        except:
            continue

               # ----------------------------
        # Métricas de classificação
        # ----------------------------

        y_pred_b = (
            y_prob_b >= cutoff
        ).astype(int)


        tn_b, fp_b, fn_b, tp_b = confusion_matrix(
            y_true_b,
            y_pred_b,
            labels=[0, 1]
        ).ravel()


        sens_boot.append(
            tp_b / (tp_b + fn_b)
            if (tp_b + fn_b) > 0 else np.nan
        )


        esp_boot.append(
            tn_b / (tn_b + fp_b)
            if (tn_b + fp_b) > 0 else np.nan
        )


        vpp_boot.append(
            tp_b / (tp_b + fp_b)
            if (tp_b + fp_b) > 0 else np.nan
        )


        vpn_boot.append(
            tn_b / (tn_b + fn_b)
            if (tn_b + fn_b) > 0 else np.nan
        )


    # ------------------------------------
    # IC95%
    # ------------------------------------

    def ic95(x):
        return np.nanpercentile(
            x,
            [2.5, 97.5]
        )


    # ------------------------------------
    # Retorno
    # ------------------------------------

    return {


        # Desempenho discriminativo
        "AUC": auc,
        "AUC_LI": ic95(aucs)[0],
        "AUC_LS": ic95(aucs)[1],


        # Calibração
        "Brier": brier,
        "Brier_LI": ic95(brier_boot)[0],
        "Brier_LS": ic95(brier_boot)[1],

        "Intercept": calibration_intercept,
        "Intercept_LI": ic95(intercept_boot)[0],
        "Intercept_LS": ic95(intercept_boot)[1],

        "Slope": calibration_slope,
        "Slope_LI": ic95(slope_boot)[0],
        "Slope_LS": ic95(slope_boot)[1],


        # Métricas clínicas
        "Sens": sens,
        "Sens_LI": ic95(sens_boot)[0],
        "Sens_LS": ic95(sens_boot)[1],

        "Esp": esp,
        "Esp_LI": ic95(esp_boot)[0],
        "Esp_LS": ic95(esp_boot)[1],

        "VPP": vpp,
        "VPP_LI": ic95(vpp_boot)[0],
        "VPP_LS": ic95(vpp_boot)[1],

        "VPN": vpn,
        "VPN_LI": ic95(vpn_boot)[0],
        "VPN_LS": ic95(vpn_boot)[1],


        # Matriz de confusão
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp
    }



# =====================================================
# Aplicar aos modelos
# =====================================================

res1 = calcular_metricas(
    modelo=xgb_novo1,
    X_test=X_test_novo1,
    y_test=y_test1,
    cutoff=cutoff1
)


res2 = calcular_metricas(
    modelo=cat_novo2,
    X_test=X_test_novo2,
    y_test=y_test2,
    cutoff=cutoff2
)


res4 = calcular_metricas(
    modelo=cat_novo4,
    X_test=X_test_novo4,
    y_test=y_test4,
    cutoff=cutoff4
)



# =====================================================
# Exibir resultados
# =====================================================

resultados = [
    ("Etapa 1", cutoff1, res1),
    ("Etapa 2", cutoff2, res2),
    ("Etapa 3", cutoff4, res4)
]


for nome, cutoff, res in resultados:


    print("\n" + "="*70)
    print(nome)
    print("="*70)


    print(
        f"Cutoff utilizado : {cutoff:.4f}\n"
    )


    print(
        f"AUC-ROC          : {res['AUC']:.4f} "
        f"({res['AUC_LI']:.4f} - {res['AUC_LS']:.4f})"
    )


    print(
        f"Sensibilidade    : {res['Sens']:.4f} "
        f"({res['Sens_LI']:.4f} - {res['Sens_LS']:.4f})"
    )


    print(
        f"Especificidade   : {res['Esp']:.4f} "
        f"({res['Esp_LI']:.4f} - {res['Esp_LS']:.4f})"
    )


    print(
        f"VPP              : {res['VPP']:.4f} "
        f"({res['VPP_LI']:.4f} - {res['VPP_LS']:.4f})"
    )


    print(
        f"VPN              : {res['VPN']:.4f} "
        f"({res['VPN_LI']:.4f} - {res['VPN_LS']:.4f})"
    )


    print("\nCalibração do modelo")


    print(
        f"Brier score      : {res['Brier']:.4f} "
        f"({res['Brier_LI']:.4f} - {res['Brier_LS']:.4f})"
    )


    print(
        f"Calibration intercept : "
        f"{res['Intercept']:.4f} "
        f"({res['Intercept_LI']:.4f} - {res['Intercept_LS']:.4f})"
    )


    print(
        f"Calibration slope     : "
        f"{res['Slope']:.4f} "
        f"({res['Slope_LI']:.4f} - {res['Slope_LS']:.4f})"
    )



    # =====================================================
    # Matriz de Confusão
    # =====================================================

    print("\nMatriz de Confusão")


    matriz = pd.DataFrame(
        [
            [res["TN"], res["FP"]],
            [res["FN"], res["TP"]]
        ],
        index=[
            "Real 0",
            "Real 1"
        ],
        columns=[
            "Predito 0",
            "Predito 1"
        ]
    )


    print(matriz)


    print("\nValores individuais:")

    print(f"TN = {res['TN']}")
    print(f"FP = {res['FP']}")
    print(f"FN = {res['FN']}")
    print(f"TP = {res['TP']}")
```

```python

# =====================================================
# Função para calcular métricas + IC95%
# =====================================================

def calcular_metricas(modelo, X_test, y_test, cutoff, n_boot=2000):

    # Probabilidades da classe positiva
    y_prob = modelo.predict_proba(X_test)[:, 1]

    # ------------------------------------
    # Métricas observadas
    # ------------------------------------
    auc = roc_auc_score(y_test, y_prob)

    y_pred = (y_prob >= cutoff).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_test,
        y_pred,
        labels=[0, 1]
    ).ravel()

    sens = tp / (tp + fn)
    esp = tn / (tn + fp)
    vpp = tp / (tp + fp) if (tp + fp) > 0 else np.nan
    vpn = tn / (tn + fn) if (tn + fn) > 0 else np.nan

    # ------------------------------------
    # Bootstrap
    # ------------------------------------
    aucs = []
    sens_boot = []
    esp_boot = []
    vpp_boot = []
    vpn_boot = []

    y_test_np = np.asarray(y_test)
    y_prob_np = np.asarray(y_prob)

    for _ in range(n_boot):

        idx = resample(
            np.arange(len(y_test_np)),
            replace=True,
            n_samples=len(y_test_np)
        )

        y_true_b = y_test_np[idx]
        y_prob_b = y_prob_np[idx]

        if len(np.unique(y_true_b)) < 2:
            continue

        aucs.append(
            roc_auc_score(y_true_b, y_prob_b)
        )

        # Usa o mesmo cutoff da etapa
        y_pred_b = (y_prob_b >= cutoff).astype(int)

        tn_b, fp_b, fn_b, tp_b = confusion_matrix(
            y_true_b,
            y_pred_b,
            labels=[0, 1]
        ).ravel()

        sens_boot.append(
            tp_b / (tp_b + fn_b)
            if (tp_b + fn_b) > 0 else np.nan
        )

        esp_boot.append(
            tn_b / (tn_b + fp_b)
            if (tn_b + fp_b) > 0 else np.nan
        )

        vpp_boot.append(
            tp_b / (tp_b + fp_b)
            if (tp_b + fp_b) > 0 else np.nan
        )

        vpn_boot.append(
            tn_b / (tn_b + fn_b)
            if (tn_b + fn_b) > 0 else np.nan
        )

    # ------------------------------------
    # IC95%
    # ------------------------------------
    def ic95(x):
        return np.nanpercentile(x, [2.5, 97.5])

    return {
        "AUC": auc,
        "AUC_LI": ic95(aucs)[0],
        "AUC_LS": ic95(aucs)[1],

        "Sens": sens,
        "Sens_LI": ic95(sens_boot)[0],
        "Sens_LS": ic95(sens_boot)[1],

        "Esp": esp,
        "Esp_LI": ic95(esp_boot)[0],
        "Esp_LS": ic95(esp_boot)[1],

        "VPP": vpp,
        "VPP_LI": ic95(vpp_boot)[0],
        "VPP_LS": ic95(vpp_boot)[1],

        "VPN": vpn,
        "VPN_LI": ic95(vpn_boot)[0],
        "VPN_LS": ic95(vpn_boot)[1]
    }

# =====================================================
# Aplicar aos 3 modelos
# =====================================================

res1 = calcular_metricas(
    modelo=xgb_novo1,
    X_test=X_test_novo1,
    y_test=y_test1,
    cutoff=cutoff1
)

res2 = calcular_metricas(
    modelo=cat_novo2,
    X_test=X_test_novo2,
    y_test=y_test2,
    cutoff=cutoff2
)

res4 = calcular_metricas(
    modelo=cat_novo4,
    X_test=X_test_novo4,
    y_test=y_test4,
    cutoff=cutoff4
)

# =====================================================
# Exibir resultados
# =====================================================

resultados = [
    ("Etapa 1", cutoff1, res1),
    ("Etapa 2", cutoff2, res2),
    ("Etapa 4", cutoff4, res4)
]

for nome, cutoff, res in resultados:

    print("\n" + "=" * 70)
    print(nome)
    print("=" * 70)

    print(f"Cutoff utilizado : {cutoff:.4f}")

    print(
        f"AUC-ROC          : {res['AUC']:.4f} "
        f"({res['AUC_LI']:.4f} - {res['AUC_LS']:.4f})"
    )

    print(
        f"Sensibilidade    : {res['Sens']:.4f} "
        f"({res['Sens_LI']:.4f} - {res['Sens_LS']:.4f})"
    )

    print(
        f"Especificidade   : {res['Esp']:.4f} "
        f"({res['Esp_LI']:.4f} - {res['Esp_LS']:.4f})"
    )

    print(
        f"VPP              : {res['VPP']:.4f} "
        f"({res['VPP_LI']:.4f} - {res['VPP_LS']:.4f})"
    )

    print(
        f"VPN              : {res['VPN']:.4f} "
        f"({res['VPN_LI']:.4f} - {res['VPN_LS']:.4f})"
    )
```
