```python
def set_seed(seed=42):
    import random
    import numpy as np
    random.seed(seed)
    np.random.seed(seed)

    try:
        import torch
        torch.manual_seed(seed)
        torch.cuda.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)
        torch.backends.cudnn.deterministic = True
        torch.backends.cudnn.benchmark = False
    except ImportError:
        pass

    try:
        import tensorflow as tf
        tf.random.set_seed(seed)
    except ImportError:
        pass
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import roc_auc_score
from sklearn.utils import resample


def auc_por_subgrupo_com_ic(X_test, y_test, y_prob, coluna, n_boot=2000, seed=42):

    df = X_test.copy()
    df = df.copy()

    df["y_true"] = np.array(y_test)
    df["y_prob"] = np.array(y_prob)

    resultados = []

    for g in df[coluna].dropna().unique():

        sub = df[df[coluna] == g]

        y_true = sub["y_true"].values
        y_prob_sub = sub["y_prob"].values

        # precisa de pelo menos 2 classes
        if len(np.unique(y_true)) < 2:
            resultados.append({
                "Subgrupo": coluna,
                "Categoria": g,
                "N": len(sub),
                "AUC-ROC": np.nan,
                "IC95%_inferior": np.nan,
                "IC95%_superior": np.nan
            })
            continue

        # AUC original
        auc_original = roc_auc_score(y_true, y_prob_sub)

        # Bootstrap
        aucs = []

        rng = np.random.default_rng(seed)  # garante reprodutibilidade SEM travar o bootstrap

        n = len(y_true)

        for _ in range(n_boot):

            idx = rng.integers(0, n, n)

            y_true_b = y_true[idx]
            y_prob_b = y_prob_sub[idx]

            # precisa de pelo menos 2 classes no bootstrap
            if len(np.unique(y_true_b)) < 2:
                continue

            aucs.append(roc_auc_score(y_true_b, y_prob_b))

        aucs = np.array(aucs)

        ic_inf = np.percentile(aucs, 2.5)
        ic_sup = np.percentile(aucs, 97.5)

        resultados.append({
            "Subgrupo": coluna,
            "Categoria": g,
            "N": len(sub),
            "AUC-ROC": auc_original,
            "IC95%_inferior": ic_inf,
            "IC95%_superior": ic_sup
        })

    return pd.DataFrame(resultados)
```

```python
auc_raca = auc_por_subgrupo_com_ic(
    X_test_novo,
    y_test,
    y_prob,
    "cs_raca_1.0"
)

print(auc_raca)
```

```python
auc_raca = auc_por_subgrupo_com_ic(
    X_test_novo,
    y_test,
    y_prob,
    "cs_raca_2.0"
)

print(auc_raca)
```

```python
auc_escola = auc_por_subgrupo_com_ic(
    X_test_novo,
    y_test,
    y_prob,
    "escola_faixa_1.0"
)

print(auc_escola)
```

```python
auc_escola = auc_por_subgrupo_com_ic(
    X_test_novo,
    y_test,
    y_prob,
    "escola_faixa_9.0"
)

print(auc_escola)
```

```python
auc_pop = auc_por_subgrupo_com_ic(
    X_test_novo,
    y_test,
    y_prob,
    "pop_grupo"
)

print(auc_pop)
```
