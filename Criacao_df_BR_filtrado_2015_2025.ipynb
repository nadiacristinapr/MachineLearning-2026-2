```python
# Criar arquivos separados por UF
# Usamos um conjunto (set) para controlar quais UFs já processamos
processados = set()

for uf in sorted(df['sg_uf'].unique()):
    uf_int = int(uf)

    # Se já processamos essa UF (no caso do 53 que vem depois do 52), pula
    if uf_int in processados:
        continue

    if uf_int in [52, 53]:
        # Filtra ambos, salva como GO_DF e marca ambos como processados
        df_uf = df[df['sg_uf'].isin([52, 53])].drop(columns=['sg_uf'])
        sigla = "GO_DF"
        processados.update([52, 53])
    else:
        # Processo normal para as outras UFs
        df_uf = df[df['sg_uf'] == uf_int].drop(columns=['sg_uf'])
        sigla = mapa_uf.get(uf_int, f"UF{uf_int}")
        processados.add(uf_int)

    # Caminho padronizado dentro da pasta criada
    nome_arquivo = f"dados{sigla}_tube_2015_2025.csv"
    df_uf.to_csv(nome_arquivo, index=False)

```
